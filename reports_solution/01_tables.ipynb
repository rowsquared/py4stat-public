# Tables: Living Conditions by Province

```python
import pandas as pd
from pathlib import Path

df = pd.read_csv(Path('..') / 'data' / '20_processed' / 'census2022_household_analysis_226.csv',
                 dtype={'QID': str})
# CSV dropped the ordered category from pd.cut: restore it so tables sort correctly
df['head_age_band'] = pd.Categorical(
    df['head_age_band'], categories=['<18', '18-34', '35-49', '50-64', '65+'], ordered=True)
```

## Task 1 — Aggregate by province

Average household size and head age by province, with a national total.

```python
summary = df.pivot_table(
    index='province_name',
    values=['DERH_HSIZE', 'DERH_HHAGE'],
    aggfunc='mean',
    margins=True,
    margins_name='National',
).round(1)
summary
```

## Task 2 — Connectivity as percentages

Row-wise percentages with `crosstab` and `normalize`.

```python
cell_by_prov = pd.crosstab(
    df['province_name'], df['H12_CELLPHONE'],
    normalize='index',
) * 100

internet_by_zone = pd.crosstab(
    df['zone'], df['has_internet'],
    normalize='index',
) * 100

cell_by_prov.round(1)
```

## Task 3 — Make it report-ready (`.style`) and label it

Format the cellphone table as one-decimal percentages with a blue gradient, then give it a label and caption so it can be cross-referenced.

```python
#| label: tbl-connectivity
#| tbl-cap: "Cellphone access by province (%)"

(cell_by_prov
 .style
 .format('{:,.1f}%')
 .background_gradient(cmap='Blues'))
```
