# Example: Load Config and Data from Catalog

This notebook shows how to use `ConfigLoader` and `DataCatalog` in **PyStatFrame** with default local path discovery.

`sys.path` is not the only option. Another common option is to install the project as editable (for example `pip install -e .`).

Here we use `sys.path` because it is simple for a standalone notebook workflow.

```python
import sys
from pathlib import Path

# In notebooks, the working directory is usually `notebooks/`.
# Add the project root so Python can import local modules such as `utilities`.
project_root = Path.cwd().resolve().parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

from utilities import ConfigLoader, DataCatalog
```

```python
# Automatically reload imported modules before each execution.
# This lets notebook changes pick up edits in `utilities/*.py` without restarting the kernel.
%load_ext autoreload
%autoreload 2
```

```python
catalog = DataCatalog("catalog.yml")
config_loader = ConfigLoader("settings.yml")

print("Available datasets:", catalog.list_datasets())
```

```python
settings = config_loader.load()
settings
```

```python
min_age = config_loader.get("cleaning.min_age")
poverty_line = config_loader.get("indicators.poverty_line")
print("min_age:", min_age)
print("poverty_line:", poverty_line)
```
