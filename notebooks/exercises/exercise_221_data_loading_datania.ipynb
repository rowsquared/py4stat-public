## Task 3 — Why loading as string matters (leading zeros)

pandas guesses data types automatically — and often gets IDs and codes wrong. Region codes like `01`, `02`, `10` can be silently converted to `1`, `2`, `10`, losing leading zeros.

Reload the file with explicit string dtypes for `hh_id` and `region_code`, then compare.

```python
# Default loading — check what region_code looks like
df_default = pd.read_csv(raw_path)
print('Default dtypes:')
print(df_default[['hh_id', 'region_code']].dtypes)
print()
print(df_default[['hh_id', 'region_code']].head(8))
```

```python
# Type-safe loading: force hh_id and region_code to string
df = pd.read_csv(
    raw_path,
)
print('Type-safe dtypes:')
print(df[['hh_id', 'region_code']].dtypes)
print()
print(df[['hh_id', 'region_code']].head(8))
```
