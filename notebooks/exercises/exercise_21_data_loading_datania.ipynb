## Task 4: Why loading as string matters (leading zeros)
*Tarefa 4: Porque é importante carregar como texto (zeros à esquerda)*

pandas guesses data types automatically, and often gets IDs and codes wrong. Region codes like `01`, `02`, `10` can be silently converted to `1`, `2`, `10`, losing leading zeros.

*O pandas adivinha os tipos de dados automaticamente e erra muitas vezes nos identificadores e códigos. Códigos de região como `01`, `02`, `10` podem ser convertidos silenciosamente para `1`, `2`, `10`, perdendo os zeros à esquerda.*

Reload the file with explicit string dtypes for `hh_id` and `region_code`, then compare.

*Volte a carregar o ficheiro com o tipo texto explícito para `hh_id` e `region_code` e depois compare.*

## Task 7: Load only the columns you need with `usecols`
*Tarefa 7: Carregar apenas as colunas necessárias com `usecols`*

With wider datasets, selecting columns at read time saves memory and speeds up loading. `usecols` tells `pd.read_csv()` to skip all other columns entirely: unlike selecting after a full load, no unnecessary data ever enters memory.

*Em conjuntos de dados mais largos, selecionar as colunas no momento da leitura poupa memória e acelera o carregamento. `usecols` diz ao `pd.read_csv()` para ignorar por completo as restantes colunas: ao contrário de selecionar depois de carregar tudo, nenhum dado desnecessário chega sequer a entrar na memória.*
