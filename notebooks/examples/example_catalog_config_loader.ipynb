# Example: Config and DataCatalog

This notebook shows how to use `Config` and `DataCatalog` from `src/utilities/`.

The pattern is:
1. load the catalog
2. inspect available datasets
3. load a dataset by logical name
4. save a result by logical name

`sys.path` is not the only option. Another common option is to install the project as editable (for example `pip install -e .`).

Here we use `sys.path` because it is simple for a standalone notebook workflow.

```python
import sys
from pathlib import Path

# In notebooks, the working directory is usually `notebooks/`.
# Add the project root so Python can find the `src/` folder.
project_root = Path.cwd()
while not (project_root / "src").exists():
    project_root = project_root.parent

sys.path.append(str(project_root))

# Import each class directly from its module file.
from src.utilities.config import Config
from src.utilities.data_catalog import DataCatalog
```

```python
# Automatically reload imported modules before each execution.
# This lets notebook changes pick up edits in `utilities/*.py` without restarting the kernel.
%load_ext autoreload
%autoreload 2
```

```python
# Step 1: load the catalog and inspect it
catalog = DataCatalog()
catalog.list_datasets()
```

```python
# Inspect the metadata for one dataset — path, format, dtypes
catalog.get("raw_households_sample")
```

```python
# Step 2 & 3: load a dataset by logical name
# The catalog resolves the path and applies dtypes / parse_dates automatically
households_df = catalog.load("raw_households_sample")
households_df.head()
```

```python
# Config: load parameters and pass them into the workflow
config = Config("parameters.yaml")

cleaning_config = config.get("cleaning")
min_age = cleaning_config.get("min_age", 15)
poverty_line = config.get("indicators", {}).get("poverty_line", 2500)

print("min_age:", min_age)
print("poverty_line:", poverty_line)
```
