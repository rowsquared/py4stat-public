## Task 1 — Classify a single age into age groups
*Tarefa 1: Classificar uma idade num grupo etário*

| Age range / Faixa etária | Label / Etiqueta |
|---|---|
| 0–14 | `Child (0-14)` |
| 15–24 | `Youth (15-24)` |
| 25–64 | `Working Age (25-64)` |
| 65+ | `Elderly (65+)` |

```python
# Task 1 — classify a single age | Tarefa 1: classificar uma idade
print("Task 1: Age Group Classification")
print("-" * 35)

age = 42  # change this value to test other cases

if age <= 14:
    age_group = "Child (0-14)"
elif age <= 24:
    age_group = "Youth (15-24)"
elif age <= 64:
    age_group = "Working Age (25-64)"
else:
    age_group = "Elderly (65+)"

print(f"Age {age} → {age_group}")
```

```python
# Task 1 (continued) — classify a list of ages | Tarefa 1 (continuação): classificar uma lista de idades
test_ages = [5, 18, 42, 70]
for test_age in test_ages:
    if test_age <= 14:
        age_group = "Child (0-14)"
    elif test_age <= 24:
        age_group = "Youth (15-24)"
    elif test_age <= 64:
        age_group = "Working Age (25-64)"
    else:
        age_group = "Elderly (65+)"
    print(f"Age {test_age} → {age_group}")
```

## Task 2 — Compound condition: programme eligibility
*Tarefa 2: Condição composta: elegibilidade para o programa*

Eligible if age is between 15 and 35 (inclusive) **and** currently unemployed.

*Elegível se a idade estiver entre 15 e 35 anos (inclusive) **e** a pessoa estiver desempregada.*

```python
# Task 2 — programme eligibility | Tarefa 2: elegibilidade para o programa
print("\nTask 2: Youth Employment Programme Eligibility")
print("-" * 45)

person_age = 28
is_employed = False

if 15 <= person_age <= 35 and not is_employed:
    eligibility = "Eligible"
else:
    eligibility = "Not Eligible"

print(f"Age: {person_age}, Employed: {is_employed}")
print(f"Youth Employment Programme: {eligibility}")
```

## Task 3 — Ternary expression for urban/rural
*Tarefa 3: Expressão ternária para urbano ou rural*

```python
# Task 3 — ternary expression | Tarefa 3: expressão ternária
print("\nTask 3: Urban/Rural Classification (Ternary)")
print("-" * 45)

is_urban = True
location_type = "Urban" if is_urban else "Rural"
print(f"Is Urban: {is_urban} → Location: {location_type}")

is_urban = False
location_type = "Urban" if is_urban else "Rural"
print(f"Is Urban: {is_urban} → Location: {location_type}")
```

## Task 4 — Batch classification with a list comprehension
*Tarefa 4: Classificação em lote com uma compreensão de lista*

```python
# Task 4 — list comprehension | Tarefa 4: compreensão de lista
print("\nTask 4: Batch Classification with List Comprehension")
print("-" * 50)

ages = [3, 16, 30, 55, 72, 12, 25, 68]
classifications = ["Child" if age < 15 else "Adult" for age in ages]

for age, cls in zip(ages, classifications):
    print(f"Age {age:2d} → {cls}")
```

**Notes:**

- Because each `elif` is only reached when the previous tests failed, `age <= 24` already implies `age >= 15` — no lower bound is needed.
- Python allows chained comparisons: `15 <= person_age <= 35` reads naturally.
- `not is_employed` is `True` when `is_employed` is `False`.
- A ternary inside a list comprehension classifies every element in one expression.

***Notas:***

- *Como cada `elif` só é alcançado quando os testes anteriores falharam, `age <= 24` já implica `age >= 15`, e não é preciso um limite inferior.*
- *O Python permite comparações encadeadas: `15 <= person_age <= 35` lê-se com naturalidade.*
- *`not is_employed` é `True` quando `is_employed` é `False`.*
- *Uma expressão ternária dentro de uma compreensão de lista classifica todos os elementos numa só linha.*
