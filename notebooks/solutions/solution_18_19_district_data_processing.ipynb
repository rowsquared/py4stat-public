# Solution 1.8–1.9 — District Data Processing

Working with lists and loops: `append()`, `enumerate()`, `zip()`, list comprehensions, `while` loops, and `range()`.

```python
# Starting data
districts = ["Numerica Urban", "Median Urban", "Polaris District", "Outlier Coastal", "Deviation District"]
target_samples = [500, 450, 300, 350, 280]
```

---

## Task 1 — Add a new district

```python
# Task 1 — append new district
print("Task 1: Adding New District")
print("-" * 35)
districts.append("Port Sample")
target_samples.append(320)
print(f"Districts: {districts}")
print(f"Targets:   {target_samples}")
```

---

## Task 2 — Print a numbered list with `enumerate()`

```python
# Task 2 — enumerate
print("\nTask 2: Numbered District List")
print("-" * 35)
for i, district in enumerate(districts, start=1):
    print(f"{i}. {district}")
```

---

## Task 3 — Total sample size with `zip()`

```python
# Task 3 — zip
print("\nTask 3: Total Sample Size")
print("-" * 40)
total_sample = 0
for district, target in zip(districts, target_samples):
    print(f"  {district}: {target}")
    total_sample += target
print(f"\nTotal target sample: {total_sample}")
```

---

## Task 4 — Filter with a list comprehension

```python
# Task 4 — list comprehension
print("\nTask 4: Districts with Target > 300")
print("-" * 40)
large_districts = [(district, target) for district, target in zip(districts, target_samples) if target > 300]
print("Large districts (target > 300):")
for item in large_districts:
    print(f"  {item}")
```

---

## Task 5 — Simulate data collection with a `while` loop

```python
# Task 5 — while loop simulation
print("\nTask 5: Data Collection Simulation")
print("-" * 40)
collected = 0
target = 100
day = 0

while collected < target:
    day += 1
    daily_collection = 20
    collected += daily_collection
    print(f"Day {day}: Collected {daily_collection}, Total: {collected}")
    if collected >= target:
        print(f"\nTarget of {target} reached in {day} days!")
        break
```

---

## Task 6 — Generate census years with `range()`

```python
# Task 6 — range
print("\nTask 6: Census Years")
print("-" * 25)
census_years = list(range(1960, 2021, 10))
print(f"Datania Census Years: {census_years}")
```

**Notes:**

- `enumerate(..., start=1)` yields `(index, value)` pairs beginning at 1 — handy for human-readable numbering.
- `zip()` walks two lists in lockstep, pairing each district with its target.
- The comprehension keeps only `(district, target)` pairs where `target > 300`.
- `range(1960, 2021, 10)` stops *before* 2021, so 2020 is the last value included.
