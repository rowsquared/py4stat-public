# Solution 1.10–1.12 — Provincial Data Structures

Choosing the right container: **tuples** for immutable reference data, **dictionaries** for structured records, and **sets** for deduplication and comparison.

## Part 1 — Tuples

## Task 1 — Create a tuple of province codes

```python
# Task 1 — province codes tuple
print("PART 1: TUPLES")
print("=" * 40)

province_codes = ("NTH", "STH", "EST", "WST", "CTR", "LKS")
print(f"Province codes: {province_codes}")

# Tuples are immutable — the line below would raise TypeError:
# province_codes[0] = "NEW"
```

## Task 2 — Unpack GPS coordinates

```python
# Task 2 — tuple unpacking
numerica_coords = (-15.4167, 28.2833)
latitude, longitude = numerica_coords
print(f"\nNumerica coordinates:")
print(f"  Latitude:  {latitude}")
print(f"  Longitude: {longitude}")
```

## Part 2 — Dictionaries

## Task 3 — Create and update a dictionary

```python
# Task 3 — dictionary
print("\nPART 2: DICTIONARIES")
print("=" * 45)

central_province = {
    "name": "Central",
    "capital": "Numerica",
    "population": 5200000,
    "area_km2": 21896
}

print(f"Province:  {central_province['name']}")
print(f"Capital:   {central_province['capital']}")
print(f"Population: {central_province['population']:,}")

# Update population
central_province["population"] = 5350000
print(f"Updated population: {central_province['population']:,}")

# Safe access for a key that might not exist
gdp = central_province.get("gdp", "Not available")
print(f"GDP: {gdp}")
```

---

## Task 4 — Access a nested dictionary

```python
# Task 4 — nested dictionary
print("\nNested Dictionary:")
print("-" * 30)
province_data = {
    "CTR": {"name": "Central",  "population": 5200000, "urban_rate": 0.75},
    "STH": {"name": "Southern", "population": 3800000, "urban_rate": 0.60},
    "NTH": {"name": "Northern", "population": 2100000, "urban_rate": 0.30},
}

southern_urban_rate = province_data["STH"]["urban_rate"]
print(f"Southern Province urban rate: {southern_urban_rate:.0%}")

for code, data in province_data.items():
    print(f"{code}: {data['name']} — Pop: {data['population']:,}")
```

## Part 3 — Sets

## Task 5 — Find unique occupations

```python
# Task 5 — sets
print("\nPART 3: SETS")
print("=" * 50)

responses = ["Farmer", "Teacher", "Farmer", "Nurse", "Teacher", "Farmer", "Driver", "Nurse"]
print(f"Raw responses ({len(responses)} total): {responses}")

unique_occupations = set(responses)
print(f"Unique occupations ({len(unique_occupations)}): {unique_occupations}")
```

## Task 6 — Compare survey rounds with set operations

```python
# Task 6 — set comparisons
print("\nComparing Survey Rounds:")
print("-" * 35)
round_1 = {"HH001", "HH002", "HH003", "HH004", "HH005"}
round_2 = {"HH003", "HH004", "HH005", "HH006", "HH007"}
print(f"Round 1: {round_1}")
print(f"Round 2: {round_2}")

panel    = round_1 & round_2   # intersection
dropped  = round_1 - round_2   # only in round 1
new      = round_2 - round_1   # only in round 2
all_hh   = round_1 | round_2   # union

print(f"\nPanel (both):        {panel}")
print(f"Dropped (round 1 only): {dropped}")
print(f"New (round 2 only):  {new}")
print(f"All unique:          {all_hh}")
```

**Notes:**

- Tuples are immutable, which makes them a good fit for fixed reference data like province codes.
- `latitude, longitude = numerica_coords` unpacks a 2-tuple into two names in one line.
- `dict.get(key, default)` avoids a `KeyError` when a key may be absent.
- Sets drop duplicates automatically; `&`, `-`, and `|` give intersection, difference, and union.
