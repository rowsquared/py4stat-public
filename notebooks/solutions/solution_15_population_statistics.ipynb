# Solution 1.5 — Population Statistics Calculator
*Solução 1.5: Calculadora de estatísticas populacionais*


---

## Task 1 — Create variables for population data
*Tarefa 1: Criar variáveis para os dados populacionais*

```python
# Task 1 — population variables | Tarefa 1: variáveis de população
total_population = 18_500_000
urban_population = 7_955_000
rural_population = 10_545_000
census_year = 2020
growth_rate = 0.0287  # 2.87 % stored as a decimal
```

---

## Task 2 — Calculate the urbanisation rate
*Tarefa 2: Calcular a taxa de urbanização*

```python
# Task 2 — urbanisation rate | Tarefa 2: taxa de urbanização
urbanisation_rate = urban_population / total_population
print(f"Urbanisation rate: {urbanisation_rate:.1%}")
```

---

## Task 3 — Project the population to 2025
*Tarefa 3: Projetar a população para 2025*

```python
# Task 3 — projected population for 2025 | Tarefa 3: população projetada para 2025
years_ahead = 2025 - census_year
projected_population_2025 = total_population * (1 + growth_rate) ** years_ahead
print(f"Projected 2025 population: {projected_population_2025:,.0f}")
```

---

## Task 4 — Print a formatted summary
*Tarefa 4: Imprimir um resumo formatado*

```python
# Task 4 — print results | Tarefa 4: imprimir resultados
print(f"Datania Population Statistics ({census_year} Census)")
print("=" * 45)
print(f"Total Population:  {total_population:>10,}")
print(f"Urban Population:  {urban_population:>10,}")
print(f"Rural Population:  {rural_population:>10,}")
print(f"Urbanisation Rate: {urbanisation_rate:>9.1%}")
print(f"Projected (2025):  {projected_population_2025:>10,.0f}")
```

---

## Task 5 — Check data types
*Tarefa 5: Verificar os tipos de dados*

```python
# Task 5 — verify types | Tarefa 5: verificar os tipos
print("\nData Type Verification:")
print(f"total_population: {type(total_population)}")
print(f"growth_rate: {type(growth_rate)}")
print(f"census_year: {type(census_year)}")
```

**Notes:**

- Underscores in numeric literals (`18_500_000`) are ignored by Python and make large numbers readable.
- `urban_population / total_population` returns a `float`; `:.1%` formats it as a percentage with one decimal.
- The compound-growth formula `current * (1 + rate) ** years` projects 5 years ahead (2020 → 2025), giving roughly 21.3 million.
- Integer literals are `int`; the decimal `growth_rate` is `float`.

***Notas:***

- *Os underscores nos números (`18_500_000`) são ignorados pelo Python e tornam os números grandes legíveis.*
- *`urban_population / total_population` devolve um `float`; `:.1%` formata como percentagem com uma casa decimal.*
- *A fórmula de crescimento composto `current * (1 + rate) ** years` projeta 5 anos (2020 para 2025), dando cerca de 21,3 milhões.*
- *Os literais inteiros são `int`; o decimal `growth_rate` é `float`.*
