# Solution 1.6 — Cleaning Survey Responses
*Solução 1.6: Limpar respostas de inquérito*

String methods for tidying inconsistent field data: `strip()`, `title()`, `lower()`, `join()`, `split()`, the `in` operator, and `replace()`.

*Métodos de texto para arrumar dados de campo inconsistentes: `strip()`, `title()`, `lower()`, `join()`, `split()`, o operador `in`, e `replace()`.*

```python
# Raw occupation data from field enumerators | Dados de ocupação em bruto vindos do terreno
raw_occupations = [
    "  subsistence FARMER  ",
    "TEACHER, secondary school",
    "  Government clerk",
    "small business OWNER  ",
    "Domestic worker "
]
```

---

## Task 1 — Clean each occupation string
*Tarefa 1: Limpar cada texto de ocupação*

```python
# Task 1 — clean occupations | Tarefa 1: limpar ocupações
print("Task 1: Cleaned Occupations")
print("-" * 30)
for occupation in raw_occupations:
    cleaned = occupation.strip().title()
    print(cleaned)
```

---

## Task 2 — Build a formatted address
*Tarefa 2: Construir uma morada formatada*

```python
# Task 2 — formatted address | Tarefa 2: morada formatada
print("\nTask 2: Formatted Address")
print("-" * 30)
address_parts = ["Plot 45", "Main Street", "Numerica", "Central Province"]
full_address = ", ".join(address_parts)
print(full_address)
```

---

## Task 3 — Extract province code from an EA code
*Tarefa 3: Extrair o código da província a partir do código da área de enumeração*

```python
# Task 3 — extract province code | Tarefa 3: extrair o código da província
print("\nTask 3: Province Code Extraction")
print("-" * 30)
ea_code = "CTR-NUM-042"
province_code = ea_code.split("-")[0]
print(f"EA Code: {ea_code}")
print(f"Province Code: {province_code}")
```

---

## Task 4 — Case-insensitive keyword search
*Tarefa 4: Pesquisa de palavra sem distinguir maiúsculas*

```python
# Task 4 — keyword search | Tarefa 4: pesquisa por palavra
print("\nTask 4: Keyword Search")
print("-" * 30)
occupation_to_check = "Subsistence Farmer"
is_farmer = "farmer" in occupation_to_check.lower()
print(f"Occupation: {occupation_to_check}")
print(f"Contains 'farmer': {is_farmer}")
```

---

## Task 5 — Replace text
*Tarefa 5: Substituir texto*

```python
# Task 5 — text replacement | Tarefa 5: substituição de texto
print("\nTask 5: Text Replacement")
print("-" * 30)
original_text = "Government clerk at Government office"
updated_text = original_text.replace("Government", "Public Sector")
print(f"Original: {original_text}")
print(f"Updated:  {updated_text}")
```

**Notes:**

- `strip()` removes leading/trailing whitespace; `title()` capitalises the first letter of each word.
- `", ".join(parts)` is the idiomatic way to build a delimited string from a list.
- `split("-")` returns a list; `[0]` takes the first segment.
- Lower-casing both sides before `in` makes the keyword search case-insensitive.
- `replace()` substitutes *every* occurrence by default.

***Notas:***

- *`strip()` remove os espaços no início e no fim; `title()` põe a primeira letra de cada palavra em maiúscula.*
- *`", ".join(parts)` é a forma idiomática de construir um texto delimitado a partir de uma lista.*
- *`split("-")` devolve uma lista; `[0]` fica com o primeiro segmento.*
- *Passar os dois lados a minúsculas antes do `in` torna a pesquisa insensível a maiúsculas.*
- *`replace()` substitui todas as ocorrências por omissão.*
