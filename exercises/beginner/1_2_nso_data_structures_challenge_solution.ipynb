```python
# Province data
province_data = [
    ("Central", 5200000, 94394),
    ("Southern", 3800000, 85283),
    ("Lakeside", 3000000, 45000),
    ("Eastern", 2500000, 51476),
    ("Northern", 2100000, 31328)
]

# Task 1: Create a dictionary from the tuples
# Keys: province names, Values: {"population": ..., "area_km2": ...}
provinces = {}
# Your code here:
for name, population, area in province_data:
    provinces[name] = {"population": population, "area_km2": area}

print("Province dictionary:", provinces)

# Task 2: Calculate population density for each province
# Add "density" key to each province's dictionary (pop/area)
# Your code here:
for name, info in provinces.items():
    population = info["population"]
    area = info["area_km2"]
    density = population / area
    provinces[name]["density"] = density


print("With density:", provinces)

# Task 3: Find provinces with density > 50 people per km²
# Your code here:
high_density_provinces = []
for name, info in provinces.items():
    if info["density"] > 50:
        high_density_provinces.append(name)

print("Provinces with density > 50:", high_density_provinces)


# Task 4: Surveyed households from two enumerators
enumerator_1 = ["HH101", "HH102", "HH103", "HH104", "HH105"]
enumerator_2 = ["HH104", "HH105", "HH106", "HH107", "HH103"]

# Find: total unique households, duplicates (visited by both), only enumerator 1
# Your code here:
unique_households = set(enumerator_1) | set(enumerator_2)
duplicates = set(enumerator_1) & set(enumerator_2)
only_enumerator_1 = set(enumerator_1) - set(enumerator_2)

print(unique_households)
print("Duplicates:", duplicates)
print("Only enumerator 1:", only_enumerator_1)
```
