```python
import pandas as pd
import datania

# Load the data
csv_file = datania.generate_census_data_with_issues(n_persons=500, seed=123)
df = pd.read_csv(csv_file)

# Check initial data types
print("Initial data types:")
print(df.dtypes)
print()

# Task 2: Ensure IDs are strings
# YOUR CODE HERE - use .astype(str)
df['person_id'] = df['person_id'].astype(str)

# Task 3: Convert age to integer (handle NaN first if needed)
# Hint: You may need to use .fillna() before .astype(int), or use pd.Int64Dtype()
# YOUR CODE HERE
df['age'] = pd.to_numeric(df['age'], errors='coerce').astype('Int64')

# Task 4: Extract province code from ea_code using .str accessor
# ea_code format is like "CTR-NUM-042" - we want "CTR"
df["province_from_ea"] = df["ea_code"].str[:3]

# Task 5: Convert urban_rural to category
# YOUR CODE HERE
df['urban_rural'] = df['urban_rural'].astype('category')

# Verify changes
print("\nFinal data types:")
print(df.dtypes)
print()
print("Sample of converted data:")
print(df[["person_id", "hh_id", "age", "province_from_ea", "urban_rural"]].head())
```
