```python

from pathlib import Path

import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt

# Download remote source files (only if missing) so the notebook can continue to load them from the local data folder.
import urllib.request

rootPath = Path().cwd().parent.parent

download_plan = [
    ("https://py4statdata.rowsquared.org/api/public/dl/XR6oc32H?inline=true",
     rootPath.joinpath("data", "power_plants", "african_power_plants.csv")),
    ("https://py4statdata.rowsquared.org/api/public/dl/OemlZbgA?inline=true",
     rootPath.joinpath("data", "geo", "africa", "africa.geojson")),
]

for url, dest_path in download_plan:
    dest_path.parent.mkdir(parents=True, exist_ok=True)
    if not dest_path.exists():
        try:
            print(f"Downloading {url} -> {dest_path}")
            # urlretrieve will follow redirects and save to the destination path
            urllib.request.urlretrieve(url, str(dest_path))
            print(f"Saved: {dest_path}")
        except Exception as e:
            print(f"Failed to download {url}: {e}")
    else:
        print(f"Already exists, skipping: {dest_path}")
```
