Option 2: Use a sample GeoTIFF

```python
import numpy as np
import rasterio
from rasterio.transform import from_bounds

# Create a sample 100x100 raster covering a small area
data = np.random.exponential(5, (100, 100)).astype(np.float32)
data[40:60, 40:60] = np.random.exponential(50, (20, 20))  # Bright city center

# Define geographic bounds (example: area in Southern Africa)
bounds = (25.0, -16.0, 26.0, -15.0)  # (west, south, east, north)
transform = from_bounds(*bounds, 100, 100)

# Write to file
with rasterio.open(
    'sample_lights.tif', 'w',
    driver='GTiff',
    height=100, width=100,
    count=1, dtype='float32',
    crs='EPSG:4326',
    transform=transform,
    nodata=-999
) as dst:
    dst.write(data, 1)

print("Created sample_lights.tif")
```
