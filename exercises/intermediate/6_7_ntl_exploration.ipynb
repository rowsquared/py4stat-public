```python
import numpy as np
import matplotlib.pyplot as plt

# Create realistic sample nighttime lights data
np.random.seed(42)
base = np.random.exponential(0.5, (200, 200))

# Add towns and cities
for _ in range(20):  # Small towns
    cx, cy = np.random.randint(20, 180, 2)
    base[cx-3:cx+3, cy-3:cy+3] += np.random.exponential(5, (6, 6))

for _ in range(3):  # Cities
    cx, cy = np.random.randint(30, 170, 2)
    base[cx-8:cx+8, cy-8:cy+8] += np.random.exponential(30, (16, 16))

lights = base.astype(np.float32)
print(f"Sample data created: {lights.shape}")

# Task 2: Explore the distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram of raw values
ax1 = axes[0]
# YOUR CODE HERE - create histogram of lights.flatten()
# Hint: use bins=100 and consider using log scale for y-axis
ax1.hist(lights.flatten(), bins=100, edgecolor='black', alpha=0.7)
ax1.set_xlabel('Light Radiance')
ax1.set_ylabel('Pixel Count')
ax1.set_title('Distribution of Nighttime Light Values')
ax1.set_yscale('log')  # Log scale helps see the tail

# Histogram of log-transformed values
ax2 = axes[1]
log_lights = np.log1p(lights)  # log(1+x) to handle zeros
# YOUR CODE HERE - histogram of log_lights
ax2.hist(log_lights.flatten(), bins=50, edgecolor='black', alpha=0.7, color='orange')
ax2.set_xlabel('Log(1 + Radiance)')
ax2.set_ylabel('Pixel Count')
ax2.set_title('Distribution of Log-Transformed Values')

plt.tight_layout()
plt.show()

# Print summary statistics
print("\n=== DISTRIBUTION SUMMARY ===")
print(f"Min: {lights.min():.2f}")
print(f"Max: {lights.max():.2f}")
print(f"Mean: {lights.mean():.2f}")
print(f"Median: {np.median(lights):.2f}")
print(f"Std: {lights.std():.2f}")

# Task 3: Classify pixels into categories
print("\n=== LIGHT CATEGORIES ===")

# Define thresholds based on typical VIIRS values
# YOUR CODE HERE - define these thresholds
thresholds = {
    'dark': (0, 0.5),      # Wilderness, water
    'dim': (0.5, 2),       # Rural areas
    'medium': (2, 10),     # Small towns
    'bright': (10, 50),    # Towns, suburbs
    'very_bright': (50, np.inf)  # Cities
}

# Task 4: Calculate percentage in each category
category_counts = {}
for category, (low, high) in thresholds.items():
    # YOUR CODE HERE - count pixels in each range
    count = # YOUR CODE HERE
    pct = count / lights.size * 100
    category_counts[category] = {'count': count, 'pct': pct}
    print(f"{category.capitalize():12} ({low:>5} - {high:>5}): {count:>8,} pixels ({pct:>5.1f}%)")

# Task 5: Visualize categories spatially
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Original image
ax1 = axes[0]
im1 = ax1.imshow(np.log1p(lights), cmap='hot')
ax1.set_title('Nighttime Lights (Log Scale)')
plt.colorbar(im1, ax=ax1)

# Classified image
ax2 = axes[1]
# Create classified array
classified = np.zeros_like(lights)
for i, (category, (low, high)) in enumerate(thresholds.items()):
    classified[(lights >= low) & (lights < high)] = i

im2 = ax2.imshow(classified, cmap='YlOrRd', vmin=0, vmax=4)
ax2.set_title('Light Categories')

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=plt.cm.YlOrRd(i/4), label=cat.capitalize())
                   for i, cat in enumerate(thresholds.keys())]
ax2.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()
```
