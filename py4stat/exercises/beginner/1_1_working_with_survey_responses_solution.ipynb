```python
# Raw occupation responses from field
raw_responses = [
    "  SUBSISTENCE farmer ",
    "teacher - PRIMARY school",
    "retail SHOP owner",
    "  Government clerk  ",
    "unemployed / seeking work"
]

# Task 1: Clean each response (strip whitespace, convert to title case)
# Your code here:
cleaned = []
for response in raw_responses:
    response = response.strip().title()
    cleaned.append(response)

print("Cleaned responses:", cleaned)

# Task 2: Create a formatted report
# Expected output format:
# Response 1: Subsistence Farmer
# Response 2: Teacher - Primary School
# (etc.)

# Your code here:
for response in cleaned:
    print(f"Response {cleaned.index(response) + 1}: {response}")

# Task 3: Find responses containing "farm" (case-insensitive)
# Your code here:
for response in cleaned:
    if "farm" in response.lower():
        print(f"Found 'farm' in: {response}")
```
